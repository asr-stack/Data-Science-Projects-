## 2. Patterns found in data (e.g. data symmetry, type of outliers, correlations etc). 

### i - Data Symmetry 

In the analysis of the Olympic winners dataset, the utilization of Kernel Density Estimate (KDE) plot is used to explore the symmetry of the data. The KDE plot is a powerful tool that helps visualization the distribution of data points, allowing to observe whether the data is symmetrically distributed around a central value.

From the KDE plot, we can observe the symmetry by examining the shape of the curve. If the curve is bell-shaped and both sides (left and right) of the peak are mirror images of each other, the data is symmetric. In the taken olympic and NOC dataset, we were to plot for the participations in the Olympics from the 1850s to 2050s , a symmetric KDE plot would indicate that the participations are contant and are evenly distributed around a central count, with not much difference in the number of participations over the years 

However, we observe that the KDE plot is skewed to the left or right, this suggests asymmetry in the data. For instance, a right-skewed KDE plot would show a longer tail on the right side, indicating that a significant number of athletes are in more number of particiations during the 1950s and till date. Similarly, a left-skewed plot would suggest a concentration of lower participations in times earlier than the 1950s. The KDE plot provides a clear visual representation of data symmetry, helping us understand the distribution characteristics of Olympic participations. Identifying whether the data is symmetric or skewed is crucial in interpreting the underlying patterns and trends within the dataset.

### ii - Types of Outliers

In the analysis we observe that the box plot is used to identify and understand the different types of outliers present in the Olympic winners dataset. Outliers are data points that lie far outside the typical range of values, and they can give us important insights or sometimes indicate errors in data collection.

The box plot helps us to easily spot these outliers. Basically, the box represents the interquartile range (IQR), which contains the middle 50% of our data. The lines, or whiskers, extend from the box to show the rest of the data, but only up to a certain limit, usually 1.5 times the IQR. Any data points beyond these whiskers are considered outliers.

There are a few types of outliers that can be observed. The most common one was the “mild outliers.” These are data points that are just outside the whiskers. They are not too far from the majority of data, but still, they are different enough to be flagged as outliers. For instance, in a box plot of athletes’ ages, a few older athletes might appear as mild outliers if most competitors are younger. Then there’s what we call “extreme outliers.” These are points that are much further from the whiskers, meaning they are significantly different from the rest of the data. For example, if we look at the number of medals won by countries, a country with a very high or very low count compared to others might be an extreme outlier.

In the plot we observe that the range of players vary from early teens to mid and late 30s that represnts our box and can be observed that there are players that come under the age group of late 40s and early nad mid 60s and 70s, these older players are concidered as to be the outliers as the number of players in the early teens and mid 30s and more in numer compared to the older players.  

Outliers in the box plot tell us a lot about the variability in our data. Mild outliers show small deviations, while extreme outliers highlight major differences or potential errors.

## 3. Common sence Insights Based on Data 

When analyzing the Olympic winners dataset, several common sense insights emerge that align with our general understanding of sports and competition.

Dominance of Certain Countries: 
It's common knowledge that some countries, like the USA, China, and Russia, tend to dominate in the Olympics. This is often reflected in the dataset, where these countries have a significantly higher medal count compared to others. This can be attributed to factors like better training facilities, a larger pool of athletes, and more resources dedicated to sports.

Age and Peak Performance: 
In many sports, athletes tend to reach their peak performance in their 20s or early 30s. The data likely shows that most medalists fall within this age range. Younger athletes might not have as much experience, while older athletes might face physical declines, so it makes sense that the majority of winners are in this "sweet spot" of athletic ability.

Gender Disparities: 
Historically, there have been differences in the participation and success rates of male and female athletes. In some countries or specific sports, male athletes might have more opportunities, leading to more medals. This could be reflected in the data, where male athletes might win more medals overall. However, in countries with more gender equality in sports, the medal counts between men and women might be more balanced.

Specialization in Certain Sports: 
Some countries are known for excelling in specific sports due to cultural or environmental factors. For example, Scandinavian countries often dominate in winter sports like skiing, while countries like Kenya and Ethiopia are renowned for their long-distance runners. The data would likely show clusters of medals in these sports for the respective countries, reflecting their specialization.

Impact of Home Host Advantage: 
Historically, the host nation of the Olympics tends to perform better, often winning more medals than in other years. This is known as the "home advantage" and could be visible in the data if you compare medal counts of host countries in different Olympic years, as the players tend to be more comfortable with theire home ground and in sence perform better.
